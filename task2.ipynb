```python
# lists

list1 = ["aaa", "bb", "c"]
list2 = list((1, 25, 33))
list3 = [i for i in range(15)]

print(list1, list2, list3, sep = "\n")
```

```python
list1.insert(1, "d")
list1.append("z")
print(list1)
list1.sort()
print(list1)
list1.sort(key = lambda x: len(x))
print(list1)
list2.extend(list1)
print(list2)
print("bb index =", list2.index("bb"))
print("list2 length =", len(list2))
```

```python
list4 = list2
list4.pop()
print(list2, list4, sep= "\n")
```

```python
import copy

list5 = copy.deepcopy(list4)
print(list5, "\n")
list5.pop()
print(list4, "\n", list5, "\n", sep = "")
list5.reverse()
print(list5)
```
